```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Hipótesis:
# H0: mu_1 = mu_2 = mu_3
# H1: al menos una media es diferente

# The error `PatsyError: Error evaluating factor: NameError: name 'horas' is not defined`
# indicates that the 'horas' column was not found by the ols function.
# This often happens if there are subtle differences in the column name, such as
# leading/trailing spaces or other invisible characters.

# To fix this, first clean all column names by stripping any leading/trailing whitespace.
df.columns = df.columns.str.strip()

# The original code had a line `df = df.rename(columns={'horas}': 'horas'})`.
# This suggests that at some point, a column named 'horas}' might have been present
# or was perceived to be the issue. Let's ensure if it exists, it's correctly renamed.
# If 'horas}' is found after stripping, rename it to 'horas'.
if 'horas}' in df.columns:
    df = df.rename(columns={'horas}': 'horas'})


modelo = ols('horas ~ C(actividad)', data=df).fit()

tabla_anova = sm.stats.anova_lm(modelo, typ=2)

tabla_anova.round(4) # round(4) es para redondear a 4 decimales.
```
